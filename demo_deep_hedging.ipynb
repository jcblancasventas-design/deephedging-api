```python
import numpy as np, pandas as pd, io

np.random.seed(123)
n = 252
dates = pd.date_range('2024-01-01', periods=n, freq='B')

def gbm(S0, mu, sigma, n):
    dt = 1/252
    return S0 * np.cumprod(np.exp((mu-0.5*sigma**2)*dt + sigma*np.sqrt(dt)*np.random.randn(n)))

df = pd.DataFrame({'SCCO': gbm(75,0.05,0.25,n), 'MINSUR': gbm(3.2,0.04,0.30,n), 'COPPER': gbm(8500,0.03,0.20,n)}, index=dates)
df.index.name = 'date'
print(df.head())
```
